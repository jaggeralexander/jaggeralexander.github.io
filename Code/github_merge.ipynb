### Merge data files for ML-based heat wave prediction

1. Import necessary packages
2. Import downloaded SST data, combine, and output for prediction
3. Example visualization of SST regression

```python
#1. Import necessary packages
#---

import os
import pandas as pd
```

```python
#2. Import downloaded/or user-created files
#---

# Define your base directory
base_dir = r"C:\Users\17753\Documents\Ch1Data"

# List of file names (modify if needed)
file_names = [
    "weather.csv",
    "250mb_gradients.csv",
    "500mb_gradients.csv",
    "850mb_gradients.csv",
    "climatology.csv",
    "gulf_sst_anom.csv",
    "mjo.csv",
    "nao.csv",
    "soi.csv",
    "soil_anomalies_full.csv"
]

# Combine the base directory with the file names
file_paths = [os.path.join(base_dir, file_name) for file_name in file_names]

# Loop through each file and display the header
for file_path in file_paths:
    try:
        # Read the CSV file
        df = pd.read_csv(file_path)
        # Print the file name and its header
        print(f"File: {file_path}")
        print("Header:", df.columns.tolist())
        print()  # Blank line for better readability
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
```
